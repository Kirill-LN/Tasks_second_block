```python
import sys
import logging
import pygame

# Настройка логирования
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')

def run_game():
    try:
        pygame.init()
        
        size = width, height = 800, 600
        speed = [2, 2]
        black = 0, 0, 0

        screen = pygame.display.set_mode(size)
        ball = pygame.Surface((50, 50), pygame.SRCALPHA)
        pygame.draw.circle(ball, (255, 0, 0), (25, 25), 25)
        ballrect = ball.get_rect()

        while True:
            for event in pygame.event.get():
                # TODO: Если тип события равен pygame.QUIT, вызовите pygame.quit() и sys.exit()
                if event.type == pygame.QUIT:
                    pygame.quit()
                    sys.exit()
                    
            ballrect = ballrect.move(speed)  # TODO: Переместите прямоугольник мяча с помощью метода .move(speed)

            # TODO: Реализуйте отскок от стен: если мяч касается левой/правой или верхней/нижней границы, инвертируйте скорость (speed[0] = -speed[0] и т.д.)
            if ballrect.left <= 0 or ballrect.right >= width:
                speed[0] = -speed[0]

            if ballrect.top <= 0 or ballrect.bottom >= height:
                speed[1] = -speed[1]
            
            
            screen.fill(black)
            # TODO: Отрисуйте мяч на экране (screen.blit) и обновите дисплей (pygame.display.flip())
            screen.blit(ball, ballrect)
            pygame.display.flip()
            
            
    except Exception as e:
        logging.error(f"Произошла ошибка: {e}")

if __name__ == "__main__":
    run_game()
```
