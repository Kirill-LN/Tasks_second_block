```python
import logging
import time
from asciimatics.screen import Screen


# Настройка логирования
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')

def demo(screen):
    try:
        for i in range(100):
            
            # TODO: Выведите текст 'Hello, Asciimatics!' по координатам (i % screen.width, screen.height // 2)
            screen.print_at(
                'Hello, Asciimatics!',
                i % screen.width,
                screen.height // 2 
            )
            # TODO: Обновите экран, чтобы отобразить изменения (метод .refresh())
            screen.refresh()
            # TODO: Сделайте паузу на 0.05 секунды с помощью time.sleep()
            time.sleep(0.05)
            # TODO: Очистите экран для следующего кадра (метод .clear())
            screen.clear()

            
        logging.info("Анимация успешно выполнена")
    except Exception as e:
        logging.error(f"Произошла ошибка: {e}")

if __name__ == "__main__":
    Screen.wrapper(demo)
```
